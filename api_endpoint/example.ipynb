```python

import requests, json
auth = 'test'
loaction = 'test'

payload = {
    'key': 'value',
}

res = requests.post(
    url=f'http://localhost:8016/api-endpoint/v1/{loaction}', 
    params={'name_search': 'test'}, 
    headers={'Authorization': auth}, 
    data=json.dumps(payload), 
)
print(res.text)
print(res.status_code)
```
